```

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# Key business metrics
print("Total transactions:", len(df))
print("Unique products:", df["Product"].nunique())
print("Unique customers:", df["Customer_ID"].nunique())
print("Total recorded sales: ₹", round(df["Total_Bill"].sum(), 2))
print("Total recorded profit: ₹", round(df["Profit"].sum(), 2))
print("Total recorded waste cost: ₹", round(df["Waste_Cost"].sum(), 2))
print("Average customer rating:", round(df["Customer_Rating"].mean(), 2))

# Top 10 products by recorded sales value
top_products = (
    df.groupby("Product")["Total_Bill"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_products.plot(kind="bar", figsize=(10, 5))
plt.title("Top 10 Products by Recorded Sales Value")
plt.xlabel("Product")
plt.ylabel("Sales value (₹)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```

import pandas as pd
import numpy as np

# 1. Understand the time period
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

print("Date range:")
print("Start:", df["Date"].min())
print("End:  ", df["Date"].max())

# 2. Check transaction structure
print("\nTransaction checks")
print("Total rows:", len(df))
print("Unique transaction IDs:", df["Transaction_ID"].nunique())
print("Repeated transaction IDs:",
      df["Transaction_ID"].duplicated().sum())

# 3. Check whether inventory figures are consistent
print("\nInventory checks")
print("Rows where Units_Produced != Units_Sold + Unsold_Units:",
      (df["Units_Produced"] !=
       df["Units_Sold"] + df["Unsold_Units"]).sum())

# 4. Inspect profit
print("\nProfit checks")
print("Negative profit rows:", (df["Profit"] < 0).sum())
print("Zero profit rows:", (df["Profit"] == 0).sum())
print("Positive profit rows:", (df["Profit"] > 0).sum())

# 5. Inspect missing values
print("\nMissing-value percentages")
missing_pct = (df.isnull().mean() * 100).sort_values(
    ascending=False
)
print(missing_pct[missing_pct > 0].round(2))

# 6. Check the available product and date coverage
print("\nUnique products:", df["Product"].nunique())
print("Unique stores:", df["Store_ID"].nunique())
print("Unique dates:", df["Date"].dt.date.nunique())
```

```

# Summarize recorded inventory and waste by product
product_analysis = (
    df.groupby("Product")
      .agg(
          records=("Transaction_ID", "count"),
          units_sold=("Units_Sold", "sum"),
          units_produced=("Units_Produced", "sum"),
          unsold_units=("Unsold_Units", "sum"),
          waste_cost=("Waste_Cost", "sum"),
          total_profit=("Profit", "sum"),
          sales_value=("Total_Bill", "sum")
      )
)

product_analysis["unsold_rate_pct"] = (
    product_analysis["unsold_units"]
    / product_analysis["units_produced"].replace(0, np.nan)
    * 100
).round(2)

product_analysis = product_analysis.sort_values(
    "waste_cost", ascending=False
)

display(product_analysis.head(10))
```

```

import pandas as pd
import numpy as np

# Prepare the date column
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# Create daily report
daily_report = (
    df.dropna(subset=["Date"])
      .groupby(df["Date"].dt.date)
      .agg(
          total_sales=("Total_Bill", "sum"),
          total_profit=("Profit", "sum"),
          total_waste_cost=("Waste_Cost", "sum"),
          transactions=("Transaction_ID", "nunique"),
          units_sold=("Units_Sold", "sum")
      )
      .reset_index(names="Date")
      .sort_values("Date")
)

# Calculate daily profit margin
daily_report["profit_margin_pct"] = (
    daily_report["total_profit"]
    / daily_report["total_sales"].replace(0, np.nan)
    * 100
).round(2)

# Identify profitable and loss-making days
daily_report["result"] = np.where(
    daily_report["total_profit"] > 0,
    "PROFIT",
    np.where(
        daily_report["total_profit"] < 0,
        "LOSS",
        "BREAK-EVEN"
    )
)

# Format monetary values
money_cols = [
    "total_sales", "total_profit", "total_waste_cost"
]
daily_report[money_cols] = daily_report[money_cols].round(2)

print("DAILY BAKERY SALES, PROFIT AND LOSS REPORT")
display(daily_report.head(15))

print("\nOVERALL DAILY PERFORMANCE")
print("Number of days:", len(daily_report))
print("Profitable days:",
      (daily_report["total_profit"] > 0).sum())
print("Loss-making days:",
      (daily_report["total_profit"] < 0).sum())
print("Break-even days:",
      (daily_report["total_profit"] == 0).sum())

print("\nHIGHEST SALES DAYS")
display(
    daily_report.nlargest(10, "total_sales")
)

print("\nBIGGEST LOSS DAYS")
display(
    daily_report.nsmallest(10, "total_profit")
)

print("\nHIGHEST PROFIT DAYS")
display(
    daily_report.nlargest(10, "total_profit")
)

# Save the complete report as a CSV file
daily_report.to_csv("bakery_daily_report.csv", index=False)
print("\nSaved: bakery_daily_report.csv")
```

```

# STEP 1: Prepare daily product demand
data = df.copy()
data["Date"] = pd.to_datetime(data["Date"], errors="coerce")
data = data.dropna(subset=["Date"])

daily_demand = (
    data.groupby(["Date", "Product"])["Quantity"]
        .sum()
        .unstack("Product", fill_value=0)
        .sort_index()
)

# Include dates with no recorded transactions for a product
all_dates = pd.date_range(
    daily_demand.index.min(),
    daily_demand.index.max(),
    freq="D"
)

daily_demand = daily_demand.reindex(
    all_dates, fill_value=0
)
daily_demand.index.name = "Date"

# STEP 2: Forecast the next 7 days
# Baseline: average demand on the same weekday
# over the previous 8 weeks.
last_date = daily_demand.index.max()
forecast_dates = pd.date_range(
    last_date + pd.Timedelta(days=1),
    periods=7,
    freq="D"
)

forecast_rows = []

for forecast_date in forecast_dates:
    history = daily_demand[
        (daily_demand.index < forecast_date) &
        (daily_demand.index.dayofweek ==
         forecast_date.dayofweek)
    ].tail(8)

    if len(history) == 0:
        history = daily_demand.tail(28)

    predicted = history.mean()

    for product, units in predicted.items():
        forecast_rows.append({
            "Forecast_Date": forecast_date.date(),
            "Product": product,
            "Predicted_Units": round(max(0, units), 2)
        })

forecast = pd.DataFrame(forecast_rows)

print("Historical data ends on:", last_date.date())
print("7-day demand forecast:")
display(forecast)

# Save the forecast
forecast.to_csv("bakery_7_day_demand_forecast.csv",
                index=False)
```
